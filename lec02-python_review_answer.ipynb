### 2-2 (슬라이드 13) 리스트 메서드와 정렬

메서드는 대부분 **원본을 바꾸고** 아무 값도 돌려주지 않음.
반면 `sorted()` 는 원본을 두고 **새 리스트를 돌려줌**.

````python
scores = [88, 95, 70]

scores.append(100)        # 끝에 추가
print(scores)

last = scores.pop()       # 마지막 원소를 꺼내며 제거
print(last, scores)

scores.sort()             # 원본을 정렬
print(scores)

new = sorted(scores)      # 새 리스트를 반환
print(new)
print(scores)             # 원본은 그대로

print(scores.append(7))   # None — 메서드는 값을 돌려주지 않음
````

### 2-6 (슬라이드 17) dict 순회와 빈도 계산

글자의 개수를 세는 코드는 거의 항상 아래 형태가 됨. 직접 써 볼 것.

````python
seq = "ATGCGATA"

freq = {}
for ch in seq:
    if ch in freq:               # 이미 나온 문자
        freq[ch] = freq[ch] + 1
    else:                        # 처음 나온 문자
        freq[ch] = 1

print(freq)

for key in freq:
    print(key, ":", freq[key])

print(sum(freq.values()))    # 값의 합계
print(list(freq.keys()))     # 키 목록
````

**연습 2-A** · 아래 서열의 염기 개수를 같은 방법으로 셀 것.

````python
seq = "ATGCGATACGATTACA"
freq = {}
for ch in seq:
    if ch in freq:
        freq[ch] = freq[ch] + 1
    else:
        freq[ch] = 1
print(freq)     # {'A': 6, 'T': 4, 'G': 3, 'C': 3}
````

### 2-7 (슬라이드 18) 집합 (set)과 집합 연산

집합은 중복을 허용하지 않으며 순서가 없음.

````python
a = {1, 2, 3, 3, 2}
b = {3, 4, 5}

print(a)          # 중복이 사라짐
print(a | b)      # 합집합
print(a & b)      # 교집합
print(a - b)      # 차집합

ids = ["s1", "s2", "s1", "s3", "s2"]
print(len(ids))            # 전체 개수
print(len(set(ids)))       # 중복을 뺀 개수
print(sorted(set(ids)))    # 순서가 필요하면 정렬해서 사용
````

**연습 2-B** · 두 실험에서 검출된 유전자 목록으로부터 공통 유전자를 구할 것.

````python
exp1 = ["TP53", "EGFR", "MYC", "TP53"]
exp2 = ["MYC", "BRCA1", "EGFR"]
common = set(exp1) & set(exp2)
print(sorted(common))   # ['EGFR', 'MYC']
````

### 2-8 (슬라이드 19) 자료구조 선택 기준

같은 문제를 list 와 dict 로 각각 풀어 보고 코드의 차이를 확인함.
학생 이름으로 점수를 찾는 상황임.

````python
names = ["kim", "lee", "park"]
scores = [88, 92, 79]

# list 방식 — 순서대로 찾아 나감
found = None
for i in range(len(names)):
    if names[i] == "lee":
        found = scores[i]
print("list 방식 :", found)

# dict 방식 — 이름으로 바로 찾음
table = {"kim": 88, "lee": 92, "park": 79}
print("dict 방식 :", table["lee"])
````

**연습 2-C** · 위 `table` 에서 80 점 이상인 학생의 이름을 모아 출력할 것.

````python
passed = []
for name in table:
    if table[name] >= 80:
        passed.append(name)
print(sorted(passed))   # ['kim', 'lee']
````

### 3-4 (슬라이드 24) 컴프리헨션 (comprehension)

리스트를 **만들기 위한 반복**은 한 줄로 줄여 쓸 수 있음.

````python
nums = [1, 2, 3, 4, 5]

# 풀어 쓴 형태
squares = []
for n in nums:
    squares.append(n * n)
print(squares)

# 한 줄로 줄인 형태
print([n * n for n in nums])

# 조건으로 걸러 내기
print([n for n in nums if n % 2 == 0])
````

### 3-5 (슬라이드 25) 함수 정의와 반환

반복해서 쓰는 절차에 이름을 붙인 것이 함수임.
`print` 는 화면에 보여 줄 뿐이므로, 값을 다른 계산에 넘기려면 `return` 이 필요함.

````python
def gc_content(seq):
    """서열에서 G 와 C 가 차지하는 비율을 돌려줌."""
    gc = seq.count("G") + seq.count("C")
    return gc / len(seq)

print(gc_content("ATGCGC"))
print(gc_content("ATAT"))
help(gc_content)
````

### 3-6 (슬라이드 26) 함수의 인자 전달 방식

````python
def report(name, score, unit="점"):
    return name + ": " + str(score) + unit

print(report("kim", 90))              # 기본값 사용
print(report("kim", 90, "%"))         # 순서대로 전달
print(report("kim", 90, unit="%"))    # 이름을 붙여 전달
print(report(score=90, name="kim"))   # 순서를 바꿔도 됨
````

### 3-7 (슬라이드 27) 변수 스코프와 기본값 주의

함수 안에서 만든 이름은 함수 밖에 남지 않음.
또한 기본값에 리스트를 쓰면 호출할 때마다 새로 만들어지지 않으므로 주의할 것.

````python
count = 0

def bump():
    count = 1        # 함수 안에서만 쓰이는 새 변수

bump()
print("바깥의 count:", count)

# 기본값에 리스트를 두면 호출할 때마다 값이 쌓임
def bad_add(x, box=[]):
    box.append(x)
    return box

print(bad_add(1))
print(bad_add(2))

# 올바른 방법
def good_add(x, box=None):
    if box is None:
        box = []
    box.append(x)
    return box

print(good_add(1))
print(good_add(2))
````

### 3-8 (슬라이드 28) 예외 처리

오류가 날 수 있는 부분을 `try` 안에 두고, 발생한 오류를 `except` 에서 처리함.

````python
def read_score(text):
    try:
        return int(text)
    except ValueError:
        print("숫자가 아님:", text)
        return None

print(read_score("90"))
print(read_score("구십"))
````

**연습 3-C** · 0 으로 나누는 경우를 처리하는 `safe_div` 를 작성할 것.

````python
def safe_div(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print("0 으로 나눌 수 없음")
        return None

print(safe_div(10, 4))
print(safe_div(10, 0))
````

### 4-1 (슬라이드 30) 클래스의 필요성 — dict 방식과의 비교

dict 로도 학생 한 명을 표현할 수 있으나, 데이터와 그것을 다루는 동작이 따로 놀게 됨.

````python
student = {"name": "kim", "scores": [90, 85]}

def average(s):
    return sum(s["scores"]) / len(s["scores"])

print(average(student))

student["scors"] = []     # 오타를 내도 오류가 나지 않고 새 키가 생김
print(list(student.keys()))
````

### 4-2 (슬라이드 31) 클래스 정의와 인스턴스 생성

`class` 는 설계도이고, 그것으로 만든 개별 객체가 인스턴스임.
`self` 는 만들어진 그 객체 자신을 가리킴.

````python
class Sample:
    def __init__(self, sid, seq):
        self.sid = sid
        self.seq = seq

    def length(self):
        return len(self.seq)

a = Sample("S01", "ATGCGC")
print(a.sid)
print(a.seq)
print(a.length())
````

**연습 4-A** · `Student` 클래스를 작성할 것. 이름과 점수 리스트를 받고, `average()` 가 평균을 돌려주도록 함.

````python
class Student:
    def __init__(self, name, scores):
        self.name = name
        self.scores = scores

    def average(self):
        return sum(self.scores) / len(self.scores)

s = Student("kim", [90, 85, 80])
print(s.name, s.average())    # kim 85.0
````

### 4-3 (슬라이드 32) 인스턴스 속성과 클래스 속성

`self.x` 는 객체마다 따로 갖는 값이고, `class` 본문에 쓴 값은 모든 객체가 함께 쓰는 값임.

````python
class Sample:
    species = "human"      # 모든 객체가 함께 쓰는 값

    def __init__(self, sid):
        self.sid = sid     # 객체마다 따로 갖는 값

a = Sample("S01")
b = Sample("S02")

print(a.sid, b.sid)
print(a.species, b.species)

Sample.species = "mouse"   # 클래스 쪽을 바꾸면 모두 바뀜
print(a.species, b.species)
````

### 4-4 (슬라이드 33) 특수 메서드 (special method)

이름이 밑줄 두 개로 감싸진 메서드는 Python 문법이 대신 불러 줌.
`__repr__` 을 정의하지 않으면 객체를 출력해도 무엇인지 알 수 없음.

````python
class Seq:
    def __init__(self, sid, seq):
        self.sid = sid
        self.seq = seq

    def __repr__(self):
        return "Seq(" + self.sid + ", len=" + str(len(self.seq)) + ")"

    def __len__(self):
        return len(self.seq)

x = Seq("S01", "ATGC")
print(x)
print(len(x))
````

### 4-5 (슬라이드 34) 상속과 메서드 오버라이딩

공통 부분은 상위 클래스에 두고, 달라지는 부분만 하위 클래스에서 다시 정의함.

````python
class Sample:
    def __init__(self, sid):
        self.sid = sid

    def describe(self):
        return "Sample " + self.sid

class RNASample(Sample):
    def __init__(self, sid, tissue):
        super().__init__(sid)      # 상위 클래스의 초기화를 먼저 부름
        self.tissue = tissue

    def describe(self):
        return "RNA " + self.sid + " / " + self.tissue

r = RNASample("S02", "liver")
print(r.describe())
print(r.sid)                   # 상위 클래스에서 물려받은 속성
````

**연습 4-B** · `Student` 를 물려받아 `GradStudent` 를 작성할 것. `advisor` 속성을 추가하고, `average()` 에 5 점을 더하도록 함.

````python
class GradStudent(Student):
    def __init__(self, name, scores, advisor):
        super().__init__(name, scores)
        self.advisor = advisor

    def average(self):
        return super().average() + 5

g = GradStudent("lee", [90, 85, 80], "류재용")
print(g.name, g.advisor, g.average())   # lee 류재용 90.0
````
